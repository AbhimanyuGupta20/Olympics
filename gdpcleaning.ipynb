```python
import pycountry as pyc
```

```python
def get_country_name(code):
    try:
        return pyc.countries.get(alpha_3=code).name
    except:
        return None
```

```python
gdp_long['Country_Name'] = gdp_long['Country_Code'].apply(get_country_name)
```

```python
# 1. Ensure your data is sorted by Country and Year 
# This is crucial for interpolation to make sense chronologically
gdp_grouped= gdp_long.sort_values(['Country_Code', 'Year'])

# 2. Group by Country and interpolate the GDP column
gdp_grouped['GDP'] = gdp_grouped.groupby('Country_Code')['GDP'].transform(lambda x: x.interpolate(method='linear'))
```

```python
gdp_grouped['GDP'] = gdp_grouped.groupby('Country_Code')['GDP'].transform(lambda x: x.interpolate().ffill().bfill())
```

```python
gdp_grouped.query('GDP.isna()')
```
