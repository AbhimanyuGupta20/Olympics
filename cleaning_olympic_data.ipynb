```python
#Drop columns for heigh and weight
athlete_data = athlete_data.drop(columns=['Height', 'Weight','Team'])
athlete_data.info()
```

```python
#Rename City column
athlete_data = athlete_data.rename(columns={"City":"GamesLocation"})
athlete_data.head()
```

```python
#Sex should be categorical
athlete_data["Sex"] = athlete_data["Sex"].astype("category")
```

```python
#replace null values in Medal to be "NoMedal" This has to happen before setting category
athlete_data['Medal'] = athlete_data["Medal"].fillna("NoMedal")
athlete_data
```

```python
#Medal data should be categorical
athlete_data["Medal"] = athlete_data["Medal"].astype("category")
```

```python
#Setting columns to be string rather than object
columns = ["Name","NOC","Games","Season","GamesLocation","Sport","Event"]
athlete_data[columns] = athlete_data[columns].astype("string")
```

```python
athlete_data.info()
```

```python
#Replaced values of Athletics with the term TrackAndField for readability. 
athlete_data['Sport'] = athlete_data['Sport'].replace('Athletics', 'TrackAndField')
athlete_data['Sport'].value_counts()
```

```python
#Remove rows where their NOC code is {'SCG', 'YUG', 'IOA', 'ROT', 'AHO'}
#these are NOCs that are not attributed to a particular country, they are either combinations of countries
#or countries that are not recognized anymore.
athlete_data = athlete_data.query('NOC != ["SCG", "YUG", "IOA", "ROT", "AHO", "KOS", "TCH"]')
athlete_data.info()
```

```python
#Change NOC in IOC codes format to be alpha 3 format - this allows us to compare with GDP data
athlete_data['NOC'] = athlete_data['NOC'].replace({"ALG": "DZA", "ANG": "AGO", "ANT": "ATG", "ARU": "ABW", "ASA": "ASM", "BAH": "BHS", "BAN": "BGD", "BAR": "BRB", "BER": "BMU", "BHU": "BTN", "BUL": "BGR", "BUR": "BFA", "CAM": "KHM", "CAY": "CYM", "CGO": "COG", "CHA": "TCD", "CHI": "CHL", "CRO": "HRV", "DEN": "DNK", "ESA": "SLV", "FIJ": "FJI", "GAM": "GMB", "GBS": "GNB", "GEQ": "GNQ", "GER": "DEU", "GRE": "GRC", "GRN": "GRD", "GUA": "GTM", "GUI": "GIN", "HAI": "HTI", "HON": "HND", "INA": "IDN", "IRI": "IRN", "ISV": "VIR", "IVB": "VGB", "KSA": "SAU", "KUW": "KWT", "LAT": "LVA", "LBA": "LBY", "LES": "LSO", "MAD": "MDG", "MAS": "MYS", "MAW": "MWI", "MRI": "MUS", "MTN": "MRT", "MYA": "MMR", "NCA": "NIC", "NED": "NLD", "NGR": "NGA", "NIG": "NER", "PAR": "PRY", "PHI": "PHL", "PLE": "PSE", "POR": "PRT", "PUR": "PRI", "RSA": "ZAF", "SLO": "SVN", "SRI": "LKA", "SUI": "CHE", "TAN": "TZA", "TGA": "TON", "TOG": "TGO", "TPE": "TWN", "UAE": "ARE", "URU": "URY", "VIE": "VNM", "VIN": "VCT", "ZAM": "ZMB", "ZIM": "ZWE", "EUN":"RUS","MGL": "MNG", "CRC": "CRI", "MON": "MCO", "LIB": "LBN", "SEY": "SYC", "BOT": "BWA", "NEP": "NPL", "SAM": "WSM", "SKN": "KNA", "SUD": "SDN", "BIZ": "BLZ", "OMA": "OMN", "VAN": "VUT", "SOL": "SLB", "BRU": "BRN"})
```

```python
#getting the code for each country
def get_country_name(code):
    try:
        return pyc.countries.get(alpha_3=code).name
    except:
        return None
```

```python
#set the name of the counries to the universal country name based on its code
#give the NOC and get the universal country name
athlete_data['CountriesRepresented'] = athlete_data['NOC'].apply(get_country_name)
athlete_data.head()
```

```python
#Set CountriesRepresented as categorical data
athlete_data["CountriesRepresented"] = athlete_data["CountriesRepresented"].astype("category")
```
